```python
"""
Predictive Coding Comparison on MNIST:
  1. Baseline: Rao & Ballard (1999) - Gaussian PC (MSE on continuous one-hot targets)
  2. Generalization: Pinchetti et al. (2022) - Categorical PC (Softmax / Cross-Entropy)

Dataset: MNIST (1x28x28 -> 784)
Hidden layers explicitly named: h1 and h2
Output: ONLY 1 Graph -> Test Accuracy (%) vs. Epochs
Colab & GPU ready.
"""

import time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

# Close any lingering plots in Colab memory
plt.close('all')

# ==========================================
# 0. Hardware & Experiment Hyperparameters
# ==========================================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[*] Running on device: {DEVICE}")

BATCH_SIZE = 128
NUM_EPOCHS = 10
INFERENCE_STEPS_T = 15     # Number of latent relaxation steps (T=15)
INFERENCE_LR_GAMMA = 0.02  # Well-calibrated relaxation step size
WEIGHT_LR = 5e-4           # Adam learning rate for synaptic weights
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(RANDOM_SEED)

# ==========================================
# 1. Dataset Preparation: MNIST (1x28x28)
# ==========================================
transform_mnist = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))  # Standard MNIST mean and std
])

train_dataset = torchvision.datasets.MNIST(
    root="./data", train=True, download=True, transform=transform_mnist
)
test_dataset = torchvision.datasets.MNIST(
    root="./data", train=False, download=True, transform=transform_mnist
)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

# ==========================================
# 2. Predictive Coding Network (784 -> 512 -> 256 -> 10)
# ==========================================
class PredictiveCodingNetwork(nn.Module):
    def __init__(self, layer_dims=[784, 512, 256, 10], variant="pinchetti"):
        super().__init__()
        self.variant = variant
        self.layer_dims = layer_dims

        # Synaptic weights and biases
        # W[0]: 784 -> 512 (Input to Hidden Layer 1)
        # W[1]: 512 -> 256 (Hidden Layer 1 to Hidden Layer 2)
        # W[2]: 256 -> 10  (Hidden Layer 2 to Output Layer)
        self.weights = nn.ParameterList([
            nn.Parameter(torch.randn(layer_dims[i], layer_dims[i-1]) * (2.0 / layer_dims[i-1])**0.5)
            for i in range(1, len(layer_dims))
        ])
        self.biases = nn.ParameterList([
            nn.Parameter(torch.zeros(layer_dims[i]))
            for i in range(1, len(layer_dims))
        ])

    def compute_energy(self, x, y_one_hot, h1, h2):
        """
        Calculates total energy F = e_1 + e_2 + e_3 using hidden states h1 and h2
        """
        # --- Hidden Layer 1 (h1) ---
        mu_h1 = F.relu(F.linear(x, self.weights[0], self.biases[0]))
        e_1 = 0.5 * torch.sum((h1 - mu_h1) ** 2, dim=-1).mean()

        # --- Hidden Layer 2 (h2) ---
        mu_h2 = F.relu(F.linear(h1, self.weights[1], self.biases[1]))
        e_2 = 0.5 * torch.sum((h2 - mu_h2) ** 2, dim=-1).mean()

        # --- Output Layer (from h2 to Output) ---
        logits = F.linear(h2, self.weights[2], self.biases[2])

        if self.variant == "pinchetti":
            # Pinchetti et al. (2022): Softmax / Categorical Cross-Entropy
            log_p = F.log_softmax(logits, dim=-1)
            e_3 = -torch.sum(y_one_hot * log_p, dim=-1).mean()
        else:
            # Rao & Ballard (1999): Gaussian MSE on continuous targets
            mu_3 = logits
            e_3 = 0.5 * torch.sum((y_one_hot - mu_3) ** 2, dim=-1).mean()

        return e_1 + e_2 + e_3

    def inference_and_update(self, x, y_one_hot, T=15, gamma=0.02, optimizer=None):
        # 1. Forward initialization: Compute initial guesses for h1 and h2
        with torch.no_grad():
            mu_h1 = F.relu(F.linear(x, self.weights[0], self.biases[0]))
            mu_h2 = F.relu(F.linear(mu_h1, self.weights[1], self.biases[1]))
            h1 = mu_h1.clone().detach().requires_grad_(True)
            h2 = mu_h2.clone().detach().requires_grad_(True)

        # 2. Inference Phase: Relax hidden states h1 and h2 to minimize total energy
        for _ in range(T):
            energy = self.compute_energy(x, y_one_hot, h1, h2)
            grad_h1, grad_h2 = torch.autograd.grad(energy, [h1, h2], retain_graph=False)

            with torch.no_grad():
                h1 -= gamma * grad_h1
                h2 -= gamma * grad_h2

            h1.requires_grad_(True)
            h2.requires_grad_(True)

        # 3. Weight Update Phase: Local Hebbian update with settled h1 and h2
        optimizer.zero_grad()
        settled_energy = self.compute_energy(x, y_one_hot, h1.detach(), h2.detach())
        settled_energy.backward()
        optimizer.step()

        return settled_energy.item()

    @torch.no_grad()
    def evaluate(self, data_loader):
        correct = 0
        total = 0
        for images, labels in data_loader:
            images = images.view(images.size(0), -1).to(DEVICE)
            labels = labels.to(DEVICE)

            h1 = F.relu(F.linear(images, self.weights[0], self.biases[0]))
            h2 = F.relu(F.linear(h1, self.weights[1], self.biases[1]))
            logits = F.linear(h2, self.weights[2], self.biases[2])

            predictions = torch.argmax(logits, dim=-1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

        return 100.0 * correct / total

# ==========================================
# 3. Training Loop
# ==========================================
def train_experiment(variant_name):
    print(f"\n{'='*25} Training MNIST [{variant_name.upper()}] {'='*25}")
    model = PredictiveCodingNetwork(layer_dims=[784, 512, 256, 10], variant=variant_name).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=WEIGHT_LR)

    accuracy_history = []

    start_time = time.time()
    for epoch in range(1, NUM_EPOCHS + 1):
        model.train()
        for images, labels in train_loader:
            images = images.view(images.size(0), -1).to(DEVICE)
            y_one_hot = F.one_hot(labels, num_classes=10).float().to(DEVICE)

            model.inference_and_update(
                images, y_one_hot, T=INFERENCE_STEPS_T, gamma=INFERENCE_LR_GAMMA, optimizer=optimizer
            )

        test_acc = model.evaluate(test_loader)
        accuracy_history.append(test_acc)
        print(f"Epoch [{epoch:02d}/{NUM_EPOCHS:02d}] | Test Accuracy: {test_acc:.2f}%")

    elapsed = time.time() - start_time
    print(f"[*] Done with {variant_name.upper()} in {elapsed:.1f}s")
    return accuracy_history

# ==========================================
# 4. Run Both Models & Display ONLY Test Accuracy Plot
# ==========================================
# 1. Baseline: Rao & Ballard (1999) - Gaussian PC
acc_gaussian = train_experiment("gaussian")

# 2. Generalization: Pinchetti et al. (2022) - Categorical / Softmax PC
acc_pinchetti = train_experiment("pinchetti")

# 3. ONLY ONE PLOT: Test Accuracy (%) vs. Epochs
epochs_range = list(range(1, NUM_EPOCHS + 1))

plt.close('all')  # Explicitly closes any old figures
fig, ax = plt.subplots(figsize=(8, 5.5), dpi=150)

ax.plot(
    epochs_range, acc_gaussian,
    marker='o', markersize=7, color='#d9534f', linewidth=2.5,
    label="Rao & Ballard (1999) [Gaussian MSE]"
)
ax.plot(
    epochs_range, acc_pinchetti,
    marker='s', markersize=7, color='#0275d8', linewidth=2.5,
    label="Pinchetti et al. (2022) [Categorical CE]"
)

ax.set_title("MNIST: Test Accuracy (%) vs. Epochs", fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel("Epoch", fontsize=12, labelpad=8)
ax.set_ylabel("Test Accuracy (%)", fontsize=12, labelpad=8)
ax.set_xticks(epochs_range)
ax.legend(frameon=True, facecolor='white', framealpha=0.9, fontsize=11, loc='lower right')
ax.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
